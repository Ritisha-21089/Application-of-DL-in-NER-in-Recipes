```python
import numpy as np
import pandas as pd

def generate_labels(row):
    """
    Generate labels for each token in the ingredient phrase based on the columns.
    """
    # Tokenize and lowercase the ingredient phrase
    tokens = str(row['Ingredient_Phrases']).lower().split()
    labels = ['O'] * len(tokens)  # Initialize all tokens with 'O'
    
    # Create a mapping of column values to their labels
    label_mapping = {
        'Quantity': 'QUANTITY',
        'Unit': 'UNIT',
        'Ingredient Name': 'NAME',
        'Form': 'FORM',
        'State': 'STATE',
        'Dry/Fresh': 'DF',
        'Size': 'SIZE',
        'Preprocessing': 'PREPROCESSING'
    }
    
    # Iterate over each column to assign labels
    for col, label in label_mapping.items():
        if pd.notna(row[col]):  # If the value exists
            value_tokens = str(row[col]).lower().split()  # Tokenize and lowercase the column value
            for i, token in enumerate(tokens):
                if token in value_tokens:
                    labels[i] = label
    
    return labels

# Apply the label generation function
df['Labels'] = df.apply(generate_labels, axis=1)
```
