```python
import pandas as pd
import numpy as np

# Assuming df is the dataframe you have already loaded with pandas

# Function to generate labels for tokens in Ingredient_Phrases
def label_tokens(row):
    # Split the Ingredient_Phrases into tokens
    tokens = str(row['Ingredient_Phrases']).split()
    labels = []

    # Check if each token appears in any other column except 'Ingredient_Phrases'
    for token in tokens:
        found = False
        for col in df.columns:
            if col != 'Ingredient_Phrases' and pd.notna(row[col]) and token in str(row[col]):
                labels.append(col)
                found = True
                break
        if not found:
            labels.append('O')  # 'O' for no match
    return tokens, labels

# Apply the function to each row
df['Tokens_and_Labels'] = df.apply(label_tokens, axis=1)

# Create separate columns for Tokens and Labels
df[['Tokens', 'Labels']] = pd.DataFrame(df['Tokens_and_Labels'].tolist(), index=df.index)

# Drop the intermediate column
df = df.drop('Tokens_and_Labels', axis=1)

# Show the dataframe with the new Tokens and Labels columns
df.head()
```
